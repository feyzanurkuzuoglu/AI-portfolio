### MATRIX MULTIPLICATION (dot product)

One of the most common errors in deep learning : shape errors

There are 2 main rules that performing matrix multiplicaiton needs to satisfy:
1. The **inner dimensions** must match:
* `(3,2) @ (3,2)` won't work
* `(3,2) @ (2,3)` will work
* `(2,3) @ (3,2)` will work
2. The resulting matrix has the shape of **outer dimensions**:
* `(3,2) @ (2,3)` -> `(3,3)`

```python
torch.matmul(tensor, tensor)
```

```python
tensor @ tensor
```

```python
tensor_A =  torch.tensor([[1,2],
              [3,4],
              [5,6]])
```

```python
tensor_B =  torch.tensor([[1,2],
              [3,4],
              [5,6]])
```

```python
torch.mm(tensor_A, tensor_B.T) # mm = matmul
```

### Reshaping, stacking, squeezing and unsqueezing

* Reshaping
* View - return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacing - combine multiple tensors on top of each other (vstack) or side by side (hstack)
* Squeeze - removes all `1` dimensions to a target tensor
* Unsqueeze - add a `1` dimension toa target tensor
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

```python
import torch
x = torch.arange(1., 10.)
x, x.shape
```

```python
x_reshaped = x.reshape(3,3)
x_reshaped
```

```python
z = x.view(3,3)
z, z.shape
```

```python
# changing z changes x (because a view of tensor shares the same memory as the original)
z[:,0] = 5
z, x
```

```python
x_stacked = torch.stack([x,x,x,x])
x_stacked
```

```python
x_stacked = torch.stack([x,x,x,x], dim=1)
x_stacked
```

```python
# torch.squeeze() - removes all single dimensions from a target tensor
x_reshaped = x_reshaped.reshape(1,9)
x_reshaped, x_reshaped.shape
```

```python
x_reshaped.squeeze(), x_reshaped.squeeze().size()
```

Örnek: (1, 3, 1) boyutunda bir tensör

x = torch.zeros(1, 3, 1)

Tüm 1'leri sil -> Sonuç: (3,)

y = torch.squeeze(x)

Sadece 0. boyutu sil -> Sonuç: (3, 1)

z = torch.squeeze(x, dim=0)

```python
# torch.unsqueeze - adds a single dimension to a target tensor at a specific dim (dimension)
x_reshaped.unsqueeze(dim=0), x_reshaped.unsqueeze(dim=0).shape
```

```python
x_reshaped.unsqueeze(dim=2), x_reshaped.unsqueeze(dim=2).shape
```

```python
# torch.permute - rearranges the dimensions of a target tensor in a specified order
# permute is a view of ther original, they share the same address
x_original = torch.randn(size=(224,224,3))
x_permuted = x_original.permute(2,0,1)

x_original.shape, x_permuted.shape
```

```python
x_original[0,0,0] = 15
x_original[0,0,0], x_permuted[0,0,0] # a view in pytorch
```

## PyTorch tensors & NumPy
* Numpy array -> pytorch tensor  : `torch.from_numpy(ndarray)`
* PyTorch tensor -> numpy array : `torch.Tensor.numpy()`

```python
array = np.arange(1.0, 8.0)
# warning: when converting from numpy -> pytorch, pytorch reflects numpy's default datatype of float64 unless specified otherwise
tensor = torch.from_numpy(array)
array, tensor
```

```python
array = array + 1
array, tensor
```

```python
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

```python
tensor = tensor + 1
tensor, numpy_tensor
```

 ## Reproducibility (trying to take random out of random)

 To reduce the randomness in neural networks and PyTorch comes the concepts of a **random seed** .




```python
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
random_tensor_A = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED) # her seferinde yazmak lazım
random_tensor_B = torch.rand(3,4)

random_tensor_A == random_tensor_B
```

For PyTorch, since it's capable of runnig compute on the GPU or CPU, it's best practice to setup device agnostic code:

```python
import torch
# setup device agnostic code
device = "cuda" if torch.cuda.is_available() else "cpu"
device
```
